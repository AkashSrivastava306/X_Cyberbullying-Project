```python
# Initialize a global set to store unique mentions without '@'
people = set()

def extract_mentions(text):
    global people  # Access the global set
    mentions = re.findall(r'@\w+', text)  # Find all mentions
    mentions_cleaned = {mention.lstrip('@') for mention in mentions}  # Remove '@' and store in a set
    people.update(mentions_cleaned)  # Add cleaned mentions to the global set
    return mentions_cleaned  # Return the cleaned mentions for the current text
df['clean_tweet'].apply(extract_mentions)
```

```python
#pos tagging using spacy
import spacy

# Load a smaller and faster spaCy model
nlp = spacy.load("en_core_web_sm", disable=['parser', 'ner'])  # POS tagging only


# Batch processing with optimized settings
def pos_spacy_batch(texts):
    results = []
    for doc in nlp.pipe(texts, batch_size=100, n_process=4):  # Increased batch_size, parallel processing
        results.append([(token.text, token.pos_) for token in doc])
    return results

# Apply POS tagging
df['spacy_pos_tags'] = pos_spacy_batch(df['clean_tweet'])
print(df.head())
```

```python

from nltk.corpus import wordnet
from nltk.stem import WordNetLemmatizer

# Initialize WordNetLemmatizer
lemmatizer = WordNetLemmatizer()

# Define a function to map spaCy POS to WordNet POS
def get_wordnet_pos(spacy_pos):
    if spacy_pos.startswith("N"):  # Noun
        return wordnet.NOUN
    elif spacy_pos.startswith("V"):  # Verb
        return wordnet.VERB
    elif spacy_pos.startswith("J"):  # Adjective
        return wordnet.ADJ
    elif spacy_pos.startswith("R"):  # Adverb
        return wordnet.ADV
    else:
        return None  # Other POS

# Lemmatization function
def lemmatize_with_wordnet(tweet):
    doc = nlp(tweet)
    lemmatized_words = []
    
    for token in doc:
        # Map spaCy POS to WordNet POS
        wordnet_pos = get_wordnet_pos(token.tag_)
        
        if wordnet_pos:  # If a valid WordNet POS is found
            lemmatized_words.append(lemmatizer.lemmatize(token.text, pos=wordnet_pos))
        else:
            lemmatized_words.append(token.text)  # Keep the word as is if no matching POS

    return " ".join(lemmatized_words)
```

```python
df["lemmatized_tweet"] = df["clean_tweet"].apply(lemmatize_with_wordnet)
```

Token Filtering Steps:

Short Tokens: Tokens with length smaller than min_length (default 2) are removed.


```python
def remove_single_char_words(text):
    """
    Function to remove words with length = 1 from the input text.
    
    Args:
    - text: The input text string.
    
    Returns:
    - A string with words having length = 1 removed.
    """
    # Split the text into words
    words = text.split()
    
    # Filter out words with length 1
    filtered_words = [word for word in words if len(word) > 1]
    
    # Join the filtered words back into a single string
    filtered_text = ' '.join(filtered_words)
    
    return filtered_text
```

```python
df['lemmatized_tweet']=df['lemmatized_tweet'].apply(remove_single_char_words)
```
